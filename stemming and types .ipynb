#### Stemming is the process 0f reducing the word into it word stem that affixed to. suffixes and prefixes or root of the words known as lemma. stemming is important for  natural language understanding and natural langauage processing 

```python
words = ['go','goes','gone','eat','eats','eaten','programming','programmer']
```

## Porter Stemmer

```python
from nltk.stem import PorterStemmer
```

```python
for word in words:
    print(word,"->",PorterStemmer().stem(word))
```

```python
# the word doesn't have meaning when we convert into root word and someone words not give answer on stemming
PorterStemmer().stem("congradulation")
```

## RegExpStemmer class

### NLTK has RegexpStemmer class with help of which can easily implement regular expression stemmer algorithms. it basically take single regular expression and remove the prefix or sufix matches the expression 

```python
from nltk.stem import RegexpStemmer

# $ symbol indeciate that end with remove those or if keep without symbol it check prefix and suffix on that word
# 
reg_stemmer = RegexpStemmer('ing|e$|s$|able$',min=4)
reg_stemmer.stem('eating')
```

## snow ball stemmer

### it gives the better accuracy

```python
from nltk.stem import SnowballStemmer
snowballstem = SnowballStemmer('english')
for word in words:
    print(word," -> ",snowballstem.stem(word))
```
