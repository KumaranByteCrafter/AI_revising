```python
corpus = """Hello welcome to kumaran works.
Please viewers review my code for understanding about me 
"""
```

```python
print(corpus)
```

```python
#tokenization
# convert the paragraph -> sentence
from nltk.tokenize import sent_tokenize
document = sent_tokenize(corpus)
for sent in document:
    print(sent)
```

```python
#Tokenization
# paragraph into words
from nltk.tokenize import word_tokenize
word_tokenize(corpus)
```

```python
# sentence into words
for sent in document:
    print(word_tokenize(sent))
```

```python
#making sure that puntuation are also part of the tokenizer
from nltk.tokenize import wordpunct_tokenize
wordpunct_tokenize(corpus)
```

```python
#same as word tokenize but little bit difference
from nltk.tokenize import TreebankWordTokenizer
tokenizer = TreebankWordTokenizer()
tokenizer.tokenize(corpus)
```

```python
import spacy

#download the enlgish model 
# spacy.cli.download("en_core_web_sm")

#load the english model for language processing 
nlp = spacy.load("en_core_web_sm")

doc = nlp(corpus)
sentence = []

#convert the paragraph into sentence
for sent in doc.sents:
    sentence.append(sent)
print(sentence)

#process the data

# sentence into word

print([word.text for sent in sentence for word in sent])
```
