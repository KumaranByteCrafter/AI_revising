# WordNetLemmetizer 

### Lemmetization is like stemming. the output will get after lemmetization is called 'lemma',which a root word rather than root stem , the output of the stemming, after the lemmetization ,we will get valid word that mean same thing

```python
from nltk.stem import WordNetLemmatizer

lemmetizer = WordNetLemmatizer()
# in lemmetizer has part of speech which use to specify grammer in this noun,verb.etc 
lemmetizer.lemmatize("going",pos='v')
```

```python
words = ['go','goes','gone','eat','eats','eaten','programming','programs']
for word in words:
    print(word," -> ",lemmetizer.lemmatize(word,pos='v'))
```

```python
spacy.cli.download("en_core_web_sm")
```

```python
# load the english model
word = "The striped bats are hanging on their feet and ate best fishes"

nlp = spacy.load('en_core_web_sm')
doc = nlp(word)
for wd in doc:
    print(wd,"->",wd.lemma_)
```
